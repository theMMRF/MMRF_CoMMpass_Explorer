```python
import numpy as np
import pandas as pd

data_path = 'data/'
```

```python
old_clin = pd.read_csv(data_path + 'clinical_data_cleaned.csv')

old_clin['public_id'] = old_clin['Tumor_Sample_Barcode'].str.split('_').str[:2].str.join('_')

old_clin['public_id']
```

```python
# IA22 standardized survival data
ia22_surv = pd.read_csv(data_path + 'mmrf_commpass_IA22_active_standardized_survival_20230414 1.csv')
print(ia22_surv.shape)

# Only keep the columns we need
ia22_surv_inter_dict = {
    "public_id": "public_id",
    # "ttpfs": "PFS",
    "ttcpfs": "PFS_censored",
    # "censpfs": "PFS_event",
    "ttpfs1": "PFS_1",
    "ttcpfs1": "PFS_1_censored",
    "censpfs1": "PFS_1_event",
    "ttos": "OS",
    "ttcos": "OS_censored",
    "censos": "OS_event"
}

ia22_surv_inter = ia22_surv[ia22_surv_inter_dict.keys()].copy()
ia22_surv_inter = ia22_surv_inter.rename(columns=ia22_surv_inter_dict)

ia22_surv_inter.shape
```

```python
# merge
new_clin = pd.merge(old_clin, ia22_surv_inter, how='left', on='public_id')

new_clin.shape
```

```python
new_clin.drop(columns=['public_id']).to_csv(data_path + 'clinical_data_cleaned_new_surv', index=False)
```
